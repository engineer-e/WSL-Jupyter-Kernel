```javascript
var fs = require("fs");
var path = require("path");

var plotlyPath = require.resolve("plotly-notebook-js");
plotlyPath = path.join(path.dirname(plotlyPath), "plotly.min.js");

console.log("Plotly.js path:", plotlyPath);
console.log("File exists:", fs.existsSync(plotlyPath));
```

```javascript

$$html$$ = `
<div id="sine-plot" style="width:100%;height:450px"></div>
<script src="https://cdn.plot.ly/plotly-2.35.2.min.js"></script>
<script>
(function () {
  const draw = function () {
    if (!window.Plotly) {
      document.getElementById("sine-plot").textContent =
        "Plotly.js did not load.";
      return;
    }

    const x = Array.from({length: 101}, (_, i) => i / 10);
    const y = x.map(v => Math.sin(v));

    window.Plotly.newPlot("sine-plot", [{
      x: x,
      y: y,
      type: "scatter",
      mode: "lines",
      name: "sin(x)"
    }], {
      title: "Sine Wave",
      xaxis: {title: "x (radians)"},
      yaxis: {title: "sin(x)"}
    });
  };

  if (window.Plotly) {
    draw();
  } else {
    const script = document.querySelector(
      'script[src="https://cdn.plot.ly/plotly-2.35.2.min.js"]'
    );
    if (script) {
      script.onload = draw;
      script.onerror = function () {
        document.getElementById("sine-plot").textContent =
          "Could not load Plotly.js.";
      };
    }
  }
})();
</script>
`;
```
